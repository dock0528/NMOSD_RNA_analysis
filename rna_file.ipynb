```python
import pandas as pd
import numpy as np
```

```python
ori_df=pd.read_csv('./RNA_DATA/nmosd_merge_sort_df.csv')
ori_df
```

```python
ref_df=pd.read_csv('./RNA_DATA/rna_correspondence_table_for_samples.csv')
ref_df
```

```python
ref_df.rename(columns={	'Sample name in GEO':'!Sample_title'},inplace=True)
```

```python
ref_df
```

```python
combined_df=pd.merge(ori_df,ref_df,how='outer',on='!Sample_title')
```

```python
combined_df
```

```python
combined_df_sort = combined_df.sort_values(by='Patient Number')
```

```python
combined_df_sort 
```

```python
merge_new_df=pd.concat([combined_df_sort.iloc[4:,:],combined_df_sort.iloc[0:4,:]],axis=0) #axis=0 依據row結合
```

```python
merge_new_df
```

```python
merge_new_df.to_csv("./RNA_DATA/merge_sample_id.csv",index=False)
```

```python
merge_new_df.loc[:,['Run','Patient Number']].to_csv('./RNA_DATA/sample_run_df.txt',sep=',',index=False)
```
