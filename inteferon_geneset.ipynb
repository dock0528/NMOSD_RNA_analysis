### **Each Collection I-IFN gene counts**

```python
import numpy as np
import pandas as pd
import os
folder_path = "./RNA_DATA/MsigDB_collection"
numbers={}
for file_name in os.listdir(folder_path):
    if file_name.endswith(".csv"):
        file_path = os.path.join(folder_path, file_name)
        print(f"讀取: {file_path}")
    interferon_genes_df=pd.read_csv(file_path)
    interferon_genes_df=interferon_genes_df.iloc[1:,]
    interferon_genes = interferon_genes_df.stack().unique() #stack()把每一欄變成一個series(自動去NA) #unique()取不重複 ->重複的gene只保留一次
    numbers[file_name] = len(interferon_genes)

# 轉成 DataFrame
interferon_df = pd.DataFrame.from_dict(numbers, orient='index', columns=["Gene_count"])
print(interferon_df)
```
