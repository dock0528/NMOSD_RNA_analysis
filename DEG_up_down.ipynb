```python
external_up=pd.read_csv("./RNA_DATA/NMOSD_RNA_DEGS_UP(External dataset).csv")
external_down=pd.read_csv("./RNA_DATA/NMOSD_RNA_DEGS_DOWN(External dataset).csv")
```

```python
housekeeping10_up=pd.read_csv("./RNA_DATA/NMOSD_RNA_DEGS_UP(10Control Genes Batch Effect Correction).csv")
housekeeping10_down=pd.read_csv("./RNA_DATA/NMOSD_RNA_DEGS_DOWN(10Control Genes Batch Effect Correction).csv")
```

```python
# 先把它們各自變成 set，加速查詢
set_ext_up   = set(external_up['x'])
set_ext_down = set(external_down['x'])
set_ctrl_up  = set(housekeeping10_up['x'])
set_ctrl_down= set(housekeeping10_down['x'])

# 建一張新的 DataFrame
df = pd.DataFrame({'DEG': external_with10_degs})

# 第 1 欄：external 是 up 還是 down
df['external_DEGs'] = df['DEG'].map(
    lambda g: 'up' if g in set_ext_up
              else ('down' if g in set_ext_down else 'NA')
)

# 第 2 欄：10 control DEGs 下是 up 還是 down
df['control_DEGs'] = df['DEG'].map(
    lambda g: 'up' if g in set_ctrl_up
              else ('down' if g in set_ctrl_down else 'NA')
)

print(df)
```

```python
def up_down_deg(intersect_degs,external_up,external_down,my_up,my_down):
    # 先把它們各自變成 set，加速查詢
    set_ext_up   = set(external_up['x'])
    set_ext_down = set(external_down['x'])
    set_my_up  = set(my_up['x'])
    set_my_down= set(my_down['x'])

    # 建一張新的 DataFrame
    df = pd.DataFrame({'DEG':intersect_degs })

    # 第 1 欄：external 是 up 還是 down
    df['external_DEGs'] = df['DEG'].map(
        lambda g: 'up' if g in set_ext_up
                else ('down' if g in set_ext_down else 'NA')
    )

    # 第 2 欄：10 control DEGs 下是 up 還是 down
    df['control_DEGs'] = df['DEG'].map(
        lambda g: 'up' if g in set_my_up
                else ('down' if g in set_my_down else 'NA')
    )

    print(df)
```

```python
up_down_deg(external_with10_degs,external_up,external_down,housekeeping10_up,housekeeping10_down)
```
